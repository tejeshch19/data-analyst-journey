```python
#pattern1
for i in range(0,5):
    for j in range(0,5):
        print("*",end=" ")
    print()
```

```python
#pattern2
for i in range(0,5):
    for j in range(0,i+1):
        print("*",end=" ")
    print()
```

```python
#pattern3
for i in range(0,5):
    for j in range(0,5-i):
        print("*",end=" ")
    print()
```

```python
#pattern4
for i in range(0,5):
    for k in range(0,5-i):
        print(" ",end=" ")
    for j in range(0,i+1):
        print("*",end=" ")
    print()
```

```python
#pattern5
for i in range(0,5):
    for k in range(0,i):
        print(" ",end=" ")
    for j in range(0,5-i):
        print("*",end=" ")
    print()
```

```python
#pattern6
for i in range(0,5):
    for j in range(0,5):
        if (i==0 or j==0 or i==4 or j==4):
            print("*",end=" ")  
        else:
            print(" ",end=" ")
    print()
```

```python
#pattern7
for i in range(0,5):
    for k in range(0,5-i):
        print("",end=" ")
    for j in range(0,i+1):
        print("*",end=" ")
    print()
for i in range(0,4):
    for k in range(0,i+2):
        print("",end=" ")
    for j in range(0,4-i):
        print("*",end=" ")
    print()
```
