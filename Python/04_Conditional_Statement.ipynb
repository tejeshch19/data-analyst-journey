```python
#contional statements- if, else,elif,nested-if
a=32
if(a>30):
    print(a,"is greater than 30")
print("end of code")
```

```python
#if-else
age =16
if(age >=18):
    print("you are eligible to vote")
else:
    print("you are not eligible to vote")
```

```python
#elif
age=13
if((age>=16) and (age<60)):
    print("you are eligible to watch movie")
elif(age>60):
    print("you are not eligible to watch movie due to voilence")
else:
    print("you are not eligible to watch movie age restriction")
```

```python
#nested if - if inside if
x=25
if(x>20):
    if(x>24):
        print(x,"is greater than 24")
    print("nested if condition")
    if(x==25):
        print("x == 25")
    else:
        print("x != 25")
print("end of code")
```
