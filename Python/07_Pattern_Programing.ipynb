- pattern 2

```python
for i in range(1,5):
    for j in range(1,i+1):
        print("*"," ",end="")
    print()
```

- pattern 3

```python
n=int(input("Enter a number"))

for i in range(0,n):
    for j in range(0,n-i):
        print("* ",end="")
    print()
```
