```python
#operator are symbols to perform operations between two operands
# 1. Arithmetic operator-> +,-,*,/,//,%,**
#simple calculator
print(7+8) #Addition
print(52-16) #Subtraction
print(2*20) #multiplication
print(30/3) #division -default in float value
print(30//3) #floor-division->removes the decimal value
print(40%5) #modulus operator
print(2**3) #power 
```

```python
print("*"+"10")
print("="*10)
print("=" + "10" + "-")
print("@" * 5 + "10" +"#" * 5)
print("*" + "10")
```

```python
#input in python
#input() by deafult it converts any value into string if we any other datatype we need to type cast it
#type casting means converting one datatype to another-int(),float(),str(),bool()
a=int(input())
b=int(input())
print("addition ",a+b)
print("subtraction",a-b)
print("multiplication",a*b)
print("division",a/b)
print("floor division",a//b)
print("modulus",a%b)
```

```python
#relational operator- compare two operators gives result as boolean
# =,==,>,<,>=,<=,!=

a=10
b=20
print(a==b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a!=b)
```

```python
# logical operator-And,Or,not
# logical and truth table:

# True and True ---->True
# False and True --->False
# True and False ---->False
# False and False --->False

# logical or truth table:
# True or True ---->True
# False or True --->True
# True or False ---->True
# False or False --->False

age=21
name="john"
print((age>18) and (name=="john"))
print((age>22) or (name=="john"))
print(age != name)
```

```python
#compound operator - +=,-=,*=,/=,//=,%=,**=
a=10
a+=3 #a=13
print(a)#a=13

a-=5 #a=13-5 =8
print(a)#a=8

a*=10#a=10*8=80
print(a)#a=80

a/=5#a=80/5=16.0
print(a)#a=16.0

a%=2#a=16.0%2=0
print(a)#a=0

a**=2#a=0**2=0
print(a)#a=0
```

```python
#floor division
a=10
a//=2#a=10//2=5
print(a)#a=5
```
