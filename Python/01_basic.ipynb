```python
# In python after print executes the cursor goes to the next line, after end stmt written, the cursor will not goes to the next line
# end="" - it prints in same line 
print("hello",end="")
print("world")

# end="\n" - it prints in next line 
print("hello",end="\n")
print("world")


# end=" " -it prints in same line with a space
print("hello",end=" ")
print("world")
```
