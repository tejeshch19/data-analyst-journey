##### operators are used to perform operations on operand or a variable

##### Types of operators
+ arithmetic operator: +,-,*,/,//,%,**


```python
a=float(input("enter the first number"))
b=float(input("enter the second number"))

addition= a+b
subtraction= a-b
multiplication= a*b
division= a/b
floorDivsion= a//b
exponential= a**b
modulus=a%b

print("addition:",addition)
print("subtraction:",subtraction)
print("multiply:",multiplication)
print("division:",division)
print("floor division:",floorDivsion)
print("modulus:",modulus)
print("exponential:",exponential)

```

- Assignment operator: =,+=,-=,*=,/=,//=,**=,%=

```python
a1=int(input("enter a number"))
a1+=3
print(a1)
a1-=4
print(a1)
a1*=5
print(a1)
a1/=44
print(a1)
a1//=6
print(a1)
a1**=20
print(a1)
a1%=80
print(a1)
```

- comparision operator: ==,>,<,>=,<=,!=

```python
a=10
b=25

print(a==b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a!=b)
```

- logical operator: and,or,not

```python
a=True
b=False

print(a and b)
print(a or b)
print(not a)
```
