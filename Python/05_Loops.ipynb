```python
#loops -while and for
#1. while loop
#syntax-
# initialization
#condition:
#  code
# updation

x=0#intialization
while x<10:#condition
    x+=1#updation
    if x==3:
        continue
    if x==7:
        break
    if x%2==0:
        pass
    print(x)
else:#clause-this else part will execute in while block has no break statement
    print("loop ends without")
print("end of code")
```
