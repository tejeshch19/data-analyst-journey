variables and DataTypes

```python
# variables are containers to store the values 
# datatype in python is define the type of value stored in variable
# types of data types
# 1.Integer
num=1253 #num is a variable, type is integer datatype
print(num)
```

```python
# 2.float -to store decimal numbers
height=6.3
print(height)
```

```python
# 3.complex-it is stores number with real and imaginary parts
com=complex(8,3)
print(com)
```

```python
# 4.Dictionary-it stores the data in key value pairs
dict={
    "name" :"tejesh",
    "group": "cse"
    # key    value
}
print(dict["name"])
print(dict["group"])
```

```python
# 5. Boolean- true or false
b1=True
b2=False
print(b1)
print(b2)
```

```python
# 6.list-it stores the multiple values in a single variable
list=[1,1,2,4,4,5]
print(list)
```

```python
# 7.set-it removes the duplicate values in list 
set1=set(list)
print(set1)
```

```python
#8. tuple- it same like list it stores in multiple values in single variable but it is immutable
tup=(
    (1,5),(2,6),(3,7)
)
print(tup)
```

```python
# 9. String-it stores collection of characters
str="HelloWorld"
print(str)
```
