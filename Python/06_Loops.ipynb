##### Loops- for,While


```python
for i in range(5):
    print(i)
```

```python
for i in range(1,8,1):
    print(i)
```

```python
for i in range(8,0,-1):
    print(i)
```

```python
for i in range(8,0,-1):
    if i==5:
        break #stops the loop if the condition match 
    print(i)
```

```python
for i in range(8,0,-1):
    if i==5:
        continue 
     #if the condition match skips that line move to next step
    print(i)
```

```python
#while loop
i=0
while i<6:
    print(i)
    i+=1
```

```python
#nested loops -a loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={i} ")
```

```python
#calculating first n natural numbers
n=10
total=0
count=1

while count<=n:
    total +=count
    count +=1
print(f"total first sum of {n} natural number {total}")
```

```python
total=0
for i in range(1,n+1,1):
    total+=i
print(f"total first sum of {n} natural number {total}")
```
